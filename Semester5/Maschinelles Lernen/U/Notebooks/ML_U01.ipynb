**Aufgabe 2: Broadcasting**

Gegeben ist eine Matrix `X` mit zufälligen Werten. Normalisieren Sie `X` **zeilenweise** (sog. Standardisierung/Z-Normalisierung), sodass jede Zeile Mittelwert 0 und Standardabweichung 1 hat, und zwar **ohne Python-Schleife über die Zeilen**, sondern ausschließlich mit Broadcasting.

Tipp: `X.mean(axis=1, keepdims=True)` und `X.std(axis=1, keepdims=True)`.

```python
np.random.seed(42)
X = np.random.uniform(0, 10, size=(4, 5))
print(X)
```

```python
### Dein Code kommt hierhin:
X_norm = (X-X.mean(axis=1,keepdims=1))/X.std(axis=1, keepdims=1)
##########################

print(X_norm)
print("Zeilenmittel nach Normalisierung:", X_norm.mean(axis=1))
print("Zeilen-Std nach Normalisierung:  ", X_norm.std(axis=1))
```

**Aufgabe 3: Vektorisierung vs. Schleife**

Implementieren Sie die paarweisen euklidischen Distanzen zwischen den Zeilen zweier Matrizen `A` (Form `(n, d)`) und `B` (Form `(m, d)`) auf zwei Arten:

a) `pairwise_distances_loop(A, B)`: mit zwei verschachtelten Python-Schleifen. Ergebnis: Matrix `D` der Form `(n, m)` mit $D_{ij} = \lVert A_i - B_j \rVert_2$.

b) `pairwise_distances_vectorized(A, B)`: vollständig vektorisiert mit NumPy (ohne Python-Schleife über die Datenpunkte). Tipp: Broadcasting über eine zusätzliche Achse, `A[:, None, :] - B[None, :, :]`.

c) Prüfen Sie mit `np.allclose`, dass beide Varianten dasselbe Ergebnis liefern, und vergleichen Sie ihre Laufzeit mit `%timeit` auf ausreichend großen Matrizen (z.B. `n=m=300`, `d=20`). Halten Sie den Laufzeitunterschied in einer Markdown-Zelle fest.

```python
def pairwise_distances_loop(A, B):
    n, m = A.shape[0], B.shape[0]
    D = np.zeros((n, m))
    ### Dein Code kommt hierhin:
    for i in range(n):
        for j in range(m):
            diff = A[i] - B[j]
            D[i,j] = np.sqrt(np.sum(diff**2))

    ##########################
    return D


def pairwise_distances_vectorized(A, B):
    ### Dein Code kommt hierhin:
    diff = A[:,None,:] - B[None,:,:]
    D = np.sqrt(np.sum(diff**2, axis=-1))
    ##########################
    return D
```

```python
np.random.seed(0)
A_small = np.random.randn(5, 3)
B_small = np.random.randn(4, 3)

D_loop = pairwise_distances_loop(A_small, B_small)
D_vec = pairwise_distances_vectorized(A_small, B_small)
print(np.allclose(D_loop, D_vec))
```

```python
np.random.seed(0)
A = np.random.randn(300, 20)
B = np.random.randn(300, 20)

%timeit pairwise_distances_loop(A, B)
%timeit pairwise_distances_vectorized(A, B)
```

*Beobachtung:* Vectorized bedeutend schneller

**Aufgabe 4: Reproduzierbarkeit**

a) Setzen Sie den Seed auf `42`, erzeugen Sie ein zufälliges Array `r1` der Form `(5,)` mit `np.random.randn`.

b) Setzen Sie den Seed erneut auf `42` und erzeugen Sie ein zweites Array `r2` genauso wie `r1`.

c) Prüfen Sie mit `np.array_equal`, dass `r1` und `r2` identisch sind.

d) Erzeugen Sie **ohne** erneutes Setzen des Seeds ein drittes Array `r3` und vergleichen Sie es mit `r1`. Was stellen Sie fest, und warum?

```python
### Dein Code kommt hierhin:
np.random.seed(42)
r1 = np.random.randn(5,)
np.random.seed(42)
r2 = np.random.randn(5,)
r3 = np.random.randn(5,)
##########################

print("r1:", r1)
print("r2:", r2)
print("r3:", r3)
print("r1 == r2:", np.array_equal(r1, r2))
print("r1 == r3:", np.array_equal(r1, r3))
```

*Antwort:* R1 und R2 sind gleich da der Generator mit der gleichen Initialisierung in die generierungen gestartet ist. Während dieser Generierung von Zahlen wird auch der Zustand aktualisiert, um als nächstes eine neue Zufallszahl zu generieren. Dementsprechend ist dieser Zustand nach der Generierung anders als vor der Generierung von r2 und sorgt für eine anderes array bei einer weiteren erzeugung
